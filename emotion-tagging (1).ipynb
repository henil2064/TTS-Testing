```python
import os

import torch
import soundfile as sf
from transformers import AutoTokenizer, AutoModelForCausalLM
from miocodec import MioCodecModel, load_audio


# ----------------------------------------------------------------------------
# Config
# ----------------------------------------------------------------------------
MODEL_NAME    = "SPRINGLab/Indic-Mio"
CODEC_NAME    = "Aratako/MioCodec-25Hz-24kHz"
REFERENCE_WAV = "/kaggle/input/datasets/henil2132/sample/WhatsApp Ptt 2026-08-19 at 12.46.33.wav"          # <-- your speaker clip (voice to use)
OUT_DIR       = "indic_mio_samples"

# How the LLM packs audio codes into its token stream (from the model card).
# Generated tokens in [SPEECH_OFFSET, SPEECH_OFFSET + NUM_AUDIO_CODES) are the
# codec's content-token indices; subtract the offset to get [0, NUM_AUDIO_CODES).
SPEECH_OFFSET   = 151669
NUM_AUDIO_CODES = 12800

INDIC_EMOTIONS   = ["happy", "sad", "angry", "disgust", "fear", "surprise"]
ENGLISH_EMOTIONS = ["happy", "sad", "enunciated", "confused", "angry", "whisper"]

# (label, base_text_without_tag, emotion_list). Add your own languages here.
TESTS = [
    ("hi", "आज मैंने कुछ ऐसा देखा जिस पर मुझे यकीन ही नहीं हुआ।", INDIC_EMOTIONS),
    ("en", "I just found out what really happened last night.",   ENGLISH_EMOTIONS),
]

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"


# ----------------------------------------------------------------------------
# Loading: LLM + codec + the reference speaker embedding (computed once)
# ----------------------------------------------------------------------------
def load():
    dtype = torch.bfloat16 if DEVICE == "cuda" else torch.float32

    print(f"[load] LLM {MODEL_NAME}  device={DEVICE}")
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
    model = AutoModelForCausalLM.from_pretrained(MODEL_NAME, torch_dtype=dtype).to(DEVICE)
    model.eval()

    print(f"[load] codec {CODEC_NAME}")
    codec = MioCodecModel.from_pretrained(CODEC_NAME).eval().to(DEVICE)

    if not os.path.exists(REFERENCE_WAV):
        raise FileNotFoundError(
            f"Reference voice clip not found: {REFERENCE_WAV!r}\n"
            "Point REFERENCE_WAV at any clean ~3-10s mono speech clip. It sets the "
            "SPEAKER/voice; the <tag> in each sentence sets the EMOTION."
        )

    print(f"[load] speaker embedding from {REFERENCE_WAV}")
    ref = load_audio(REFERENCE_WAV, sample_rate=codec.config.sample_rate).to(DEVICE)
    global_embedding = codec.encode(ref, return_content=False).global_embedding  # (dim,)

    return tokenizer, model, codec, global_embedding


# ----------------------------------------------------------------------------
# Synthesis: text (+ optional emotion tag) -> mono waveform (numpy float array)
# ----------------------------------------------------------------------------
@torch.inference_mode()
def synthesize(tokenizer, model, codec, global_embedding, text, emotion=None):
    if emotion:
        text = f"{text} <{emotion}>"          # tag goes at the END of the sentence

    messages = [{"role": "user", "content": text}]
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=1024,
        do_sample=True,            # required for temperature/top_p to take effect
        temperature=0.9,
        top_p=0.9,
    )

    # Keep the newly generated tokens, then pull out the codec content tokens.
    generated = output[0][inputs["input_ids"].shape[1]:]
    audio_codes = [
        t.item() - SPEECH_OFFSET
        for t in generated
        if SPEECH_OFFSET <= t.item() < SPEECH_OFFSET + NUM_AUDIO_CODES
    ]
    if not audio_codes:
        raise RuntimeError(f"No audio tokens produced for: {text!r}")

    content_token_indices = torch.tensor(audio_codes, dtype=torch.long, device=DEVICE)  # (seq_len,)
    wav = codec.decode(
        global_embedding=global_embedding,
        content_token_indices=content_token_indices,
    )
    return wav.squeeze().float().cpu().numpy()


# ----------------------------------------------------------------------------
# Main: generate a WAV for every emotion in every test bucket
# ----------------------------------------------------------------------------
def main():
    os.makedirs(OUT_DIR, exist_ok=True)
    tokenizer, model, codec, global_embedding = load()
    sr = codec.config.sample_rate          # 24000 for this codec

    ok, failed = 0, 0
    for label, text, emotions in TESTS:
        for emotion in [None, *emotions]:          # None = neutral baseline (no tag)
            name = f"{label}_{emotion or 'neutral'}"
            print(f"\n[gen] {name}: {text} {'<'+emotion+'>' if emotion else '(no tag)'}")
            try:
                audio = synthesize(tokenizer, model, codec, global_embedding, text, emotion)
                path = os.path.join(OUT_DIR, f"{name}.wav")
                sf.write(path, audio, sr)
                print(f"      saved -> {path}  ({len(audio) / sr:.2f}s @ {sr} Hz)")
                ok += 1
            except Exception as e:                 # keep going even if one emotion fails
                print(f"      FAILED: {e}")
                failed += 1

    print(f"\n[done] {ok} ok, {failed} failed. WAVs are in ./{OUT_DIR}/")


if __name__ == "__main__":
    main()
```

# Auto tagging

```python
# ===== Cell A: automatic emotion tagger =====
from transformers import pipeline, AutoTokenizer, AutoModelForCausalLM
import torch

# any classifier/LLM label -> Indic-Mio tag  (None = neutral, no tag)
TO_TAG = {
    "joy": "happy", "happy": "happy",
    "sadness": "sad", "sad": "sad",
    "anger": "angry", "angry": "angry",
    "fear": "fear", "disgust": "disgust", "surprise": "surprise",
    "neutral": None,
    "love": "happy", "gratitude": "happy", "frustration": "angry", "contempt": "disgust",
}
_ALLOWED = {"happy", "sad", "angry", "disgust", "fear", "surprise", "neutral"}

# backend 1: multilingual classifier — fast, literal emotion, works on Hindi
_clf = pipeline("text-classification",
                model="tabularisai/multilingual-emotion-classification", top_k=1)

def _label_via_classifier(text):
    out = _clf(text)
    if isinstance(out[0], list):          # normalize [[{..}]] vs [{..}]
        out = out[0]
    return out[0]["label"].lower()

# backend 2 (optional): small open LLM — context-aware "delivery" emotion
_llm = {}
def _label_via_llm(text, model_id="Qwen/Qwen2.5-1.5B-Instruct"):
    if not _llm:
        _llm["tok"] = AutoTokenizer.from_pretrained(model_id)
        _llm["model"] = AutoModelForCausalLM.from_pretrained(
            model_id, torch_dtype=torch.bfloat16, device_map="auto")
    tok, m = _llm["tok"], _llm["model"]
    sys = ("You decide how a line should be SPOKEN ALOUD by a voice assistant. "
           "Use context, not just literal words (e.g. 'your reports are normal' -> happy/reassuring). "
           "Answer with exactly ONE word from: happy, sad, angry, disgust, fear, surprise, neutral.")
    msgs = [{"role": "system", "content": sys}, {"role": "user", "content": text}]
    prompt = tok.apply_chat_template(msgs, tokenize=False, add_generation_prompt=True)
    ids = tok(prompt, return_tensors="pt").to(m.device)
    out = m.generate(**ids, max_new_tokens=4, do_sample=False)
    ans = tok.decode(out[0][ids["input_ids"].shape[1]:], skip_special_tokens=True).strip().lower()
    ans = ans.split()[0].strip(".,!?।") if ans else "neutral"
    return ans if ans in _ALLOWED else "neutral"

def auto_emotion(text, method="classifier"):
    """text -> Indic-Mio tag (or None). method: 'classifier' | 'llm'."""
    label = _label_via_llm(text) if method == "llm" else _label_via_classifier(text)
    return TO_TAG.get(label, None)
```

# fishaudio/s2-pro

```python
import requests
from IPython.display import Audio, display

OPENROUTER_API_KEY = ""


text = (
    "Wow, this actually works! "
    "[whispering] but let me tell you a little secret... "
    "[audience laughter] just kidding! "
    "[sad] I hope you don't leave so soon."
)

response = requests.post(
    url="https://openrouter.ai/api/v1/audio/speech",
    headers={
        "Authorization": f"Bearer {OPENROUTER_API_KEY}",
        "Content-Type": "application/json",
    },
    json={
        "model": "fish-audio/s2-pro",
        "input": text,
        "response_format": "mp3",
    },
)

if response.ok and response.headers.get("content-type", "").startswith("audio"):
    with open("output.mp3", "wb") as f:
        f.write(response.content)
    print(f"✅ Saved {len(response.content)} bytes | ID: {response.headers.get('X-Generation-Id')}")
    display(Audio("output.mp3", autoplay=True))
else:
    print(f"❌ Failed ({response.status_code}):\n{response.text}")
```
