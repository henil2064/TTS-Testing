```python
import os, torch, soundfile as sf
from transformers import AutoTokenizer, AutoModelForCausalLM
from miocodec import MioCodecModel, load_audio
from IPython.display import Audio, display

MODEL_NAME      = "SPRINGLab/Indic-Mio"
CODEC_NAME      = "Aratako/MioCodec-25Hz-24kHz"
REFERENCE_WAV   = "/kaggle/working/ref_10s.wav"
SPEECH_OFFSET   = 151669
NUM_AUDIO_CODES = 12800
DEVICE          = "cuda" if torch.cuda.is_available() else "cpu"


def load():
    dtype = torch.bfloat16 if DEVICE == "cuda" else torch.float32
    print(f"[load] LLM {MODEL_NAME}  device={DEVICE}")
    tok = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
    mdl = AutoModelForCausalLM.from_pretrained(MODEL_NAME, dtype=dtype).to(DEVICE).eval()
    print(f"[load] codec {CODEC_NAME}")
    cdc = MioCodecModel.from_pretrained(CODEC_NAME).eval().to(DEVICE)
    return tok, mdl, cdc


def set_reference(path):
    """Recompute the speaker embedding from `path`. Always call this after changing the clip."""
    global REFERENCE_WAV, global_embedding
    REFERENCE_WAV = path
    ref = load_audio(path, sample_rate=codec.config.sample_rate).to(DEVICE)
    global_embedding = codec.encode(ref, return_content=False).global_embedding
    print(f"[ref] {path} | dim={tuple(global_embedding.shape)} | norm={global_embedding.norm():.4f}")
    return global_embedding


@torch.inference_mode()
def synthesize(tokenizer, model, codec, global_embedding, text,
               emotion=None, temperature=0.9, top_p=0.9):
    if emotion:
        text = f"{text} <{emotion}>"                    # tag goes at the END
    prompt = tokenizer.apply_chat_template(
        [{"role": "user", "content": text}], tokenize=False, add_generation_prompt=True)
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(**inputs, max_new_tokens=1024, do_sample=True,
                            temperature=temperature, top_p=top_p)

    generated = output[0][inputs["input_ids"].shape[1]:]
    audio_codes = [t.item() - SPEECH_OFFSET for t in generated
                   if SPEECH_OFFSET <= t.item() < SPEECH_OFFSET + NUM_AUDIO_CODES]
    if not audio_codes:
        raise RuntimeError(f"No audio tokens produced for: {text!r}")

    idx = torch.tensor(audio_codes, dtype=torch.long, device=DEVICE)
    wav = codec.decode(global_embedding=global_embedding, content_token_indices=idx)
    return wav.squeeze().float().cpu().numpy()


tokenizer, model, codec = load()
sr = codec.config.sample_rate
global_embedding = set_reference(REFERENCE_WAV)
```

```python
import os
import time
import json

import torch
import soundfile as sf
from transformers import AutoTokenizer, AutoModelForCausalLM
from miocodec import MioCodecModel, load_audio


# ----------------------------------------------------------------------------
# Config
# ----------------------------------------------------------------------------
MODEL_NAME    = "SPRINGLab/Indic-Mio"
CODEC_NAME    = "Aratako/MioCodec-25Hz-24kHz"
REFERENCE_WAV = "/kaggle/working/ref_10s.wav"          # <-- your speaker clip (voice to use)
OUT_DIR       = "indic_mio_samples"

# How the LLM packs audio codes into its token stream (from the model card).
# Generated tokens in [SPEECH_OFFSET, SPEECH_OFFSET + NUM_AUDIO_CODES) are the
# codec's content-token indices; subtract the offset to get [0, NUM_AUDIO_CODES).
SPEECH_OFFSET   = 151669
NUM_AUDIO_CODES = 12800

INDIC_EMOTIONS   = ["happy", "sad", "angry", "disgust", "fear", "surprise"]
ENGLISH_EMOTIONS = ["happy", "sad", "enunciated", "confused", "angry", "whisper"]

# (label, base_text_without_tag, emotion_list). Add your own languages here.
TESTS = [
    ("hi", "आज मैंने कुछ ऐसा देखा जिस पर मुझे यकीन ही नहीं हुआ।", INDIC_EMOTIONS),
    ("mr", "आज मी असं काहीतरी पाहिलं ज्यावर माझा विश्वासच बसला नाही.", INDIC_EMOTIONS),
    ("ta", "இன்று நான் நம்பவே முடியாத ஒன்றைப் பார்த்தேன்.", INDIC_EMOTIONS),
    ("te", "ఈరోజు నేను నమ్మలేని ఒక విషయాన్ని చూశాను.", INDIC_EMOTIONS),
    ("en", "I just found out what really happened last night.", ENGLISH_EMOTIONS),
]

DEVICE = "cuda" if torch.cuda.is_available() else "cpu"


# ----------------------------------------------------------------------------
# Loading: LLM + codec + the reference speaker embedding (computed once)
# ----------------------------------------------------------------------------
def load():
    dtype = torch.bfloat16 if DEVICE == "cuda" else torch.float32

    print(f"[load] LLM {MODEL_NAME}  device={DEVICE}")
    tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME, trust_remote_code=True)
    model = AutoModelForCausalLM.from_pretrained(MODEL_NAME, torch_dtype=dtype).to(DEVICE)
    model.eval()

    print(f"[load] codec {CODEC_NAME}")
    codec = MioCodecModel.from_pretrained(CODEC_NAME).eval().to(DEVICE)

    if not os.path.exists(REFERENCE_WAV):
        raise FileNotFoundError(
            f"Reference voice clip not found: {REFERENCE_WAV!r}\n"
            "Point REFERENCE_WAV at any clean ~3-10s mono speech clip. It sets the "
            "SPEAKER/voice; the <tag> in each sentence sets the EMOTION."
        )

    print(f"[load] speaker embedding from {REFERENCE_WAV}")
    ref = load_audio(REFERENCE_WAV, sample_rate=codec.config.sample_rate).to(DEVICE)
    global_embedding = codec.encode(ref, return_content=False).global_embedding  # (dim,)

    return tokenizer, model, codec, global_embedding


# ----------------------------------------------------------------------------
# Synthesis: text (+ optional emotion tag) -> mono waveform (numpy float array)
# ----------------------------------------------------------------------------
@torch.inference_mode()
def synthesize(tokenizer, model, codec, global_embedding, text, emotion=None):
    if emotion:
        text = f"{text} <{emotion}>"          # tag goes at the END of the sentence

    messages = [{"role": "user", "content": text}]
    prompt = tokenizer.apply_chat_template(
        messages, tokenize=False, add_generation_prompt=True
    )
    inputs = tokenizer(prompt, return_tensors="pt").to(model.device)

    output = model.generate(
        **inputs,
        max_new_tokens=1024,
        do_sample=True,            # required for temperature/top_p to take effect
        temperature=0.9,
        top_p=0.9,
    )

    # Keep the newly generated tokens, then pull out the codec content tokens.
    generated = output[0][inputs["input_ids"].shape[1]:]
    audio_codes = [
        t.item() - SPEECH_OFFSET
        for t in generated
        if SPEECH_OFFSET <= t.item() < SPEECH_OFFSET + NUM_AUDIO_CODES
    ]
    if not audio_codes:
        raise RuntimeError(f"No audio tokens produced for: {text!r}")

    content_token_indices = torch.tensor(audio_codes, dtype=torch.long, device=DEVICE)  # (seq_len,)
    wav = codec.decode(
        global_embedding=global_embedding,
        content_token_indices=content_token_indices,
    )
    return wav.squeeze().float().cpu().numpy()


# ----------------------------------------------------------------------------
# Main: generate a WAV for every emotion in every test bucket
# ----------------------------------------------------------------------------
def main():
    os.makedirs(OUT_DIR, exist_ok=True)
    tokenizer, model, codec, global_embedding = load()
    sr = codec.config.sample_rate          # 24000 for this codec

    manifest = []   # per-file text/emotion/lang, used later for WER/CER/etc.
    timings  = {}   # per-file synth time -> RTF

    ok, failed = 0, 0
    for label, text, emotions in TESTS:
        for emotion in [None, *emotions]:          # None = neutral baseline (no tag)
            name = f"{label}_{emotion or 'neutral'}"
            print(f"\n[gen] {name}: {text} {'<'+emotion+'>' if emotion else '(no tag)'}")
            try:
                t0 = time.time()
                audio = synthesize(tokenizer, model, codec, global_embedding, text, emotion)
                synth_seconds = time.time() - t0
                path = os.path.join(OUT_DIR, f"{name}.wav")
                sf.write(path, audio, sr)
                audio_seconds = len(audio) / sr
                rtf = synth_seconds / audio_seconds if audio_seconds > 0 else float("nan")
                print(f"      saved -> {path}  ({audio_seconds:.2f}s @ {sr} Hz, synth {synth_seconds:.2f}s, RTF {rtf:.2f})")

                manifest.append({"file": name, "language": label, "text": text, "emotion": emotion or "neutral"})
                timings[name] = {"synth_seconds": synth_seconds, "audio_seconds": audio_seconds, "rtf": rtf}
                ok += 1
            except Exception as e:                 # keep going even if one emotion fails
                print(f"      FAILED: {e}")
                failed += 1

    with open(os.path.join(OUT_DIR, "manifest.json"), "w", encoding="utf-8") as f:
        json.dump(manifest, f, ensure_ascii=False, indent=2)
    with open(os.path.join(OUT_DIR, "timings.json"), "w", encoding="utf-8") as f:
        json.dump(timings, f, ensure_ascii=False, indent=2)

    print(f"\n[done] {ok} ok, {failed} failed. WAVs + manifest.json + timings.json are in ./{OUT_DIR}/")


if __name__ == "__main__":
    main()
```

# Automatic evaluation: WER, CER, and everything else that would normally need a human listener

Everything below is computed with models/formulas, no human raters required. It reads the
`manifest.json` + `timings.json` that the generation cells above now write next to each
batch of WAVs, so run this section any time after generating a batch.

| Metric | What it normally needs a human for | What replaces the human here |
|---|---|---|
| **WER / CER** | A person checking "did it actually say the sentence?" | ASR transcription (Whisper) + string diff against the input text |
| **RTF (Real-Time Factor)** | N/A (already objective) | `synthesis_time / audio_duration`, logged during generation |
| **Speaker similarity** | A person judging "does this sound like the same speaker as the reference clip?" | Cosine similarity between speaker embeddings (ECAPA-TDNN) |
| **Naturalness (MOS proxy)** | MOS — a panel of humans rating naturalness 1-5 | UTMOS, a neural MOS predictor trained to imitate those human ratings |
| **Emotion match** | A person judging "does this actually sound happy / sad / angry?" | A speech-emotion-recognition (SER) classifier compared against the emotion tag you asked for |

**Caveats worth knowing before you trust the numbers:**
- The SER model used below (`superb/wav2vec2-base-superb-er`) is trained on **English** IEMOCAP data with only 4 classes (neutral/happy/angry/sad). It is *not* meaningful for the Hindi samples, and it has no notion of `disgust/fear/surprise/whisper/enunciated/confused` — those come back as `not_supported`. Treat "emotion match" as a rough sanity check for English happy/sad/angry/neutral only, not ground truth.
- UTMOS was trained mostly on English/Japanese listening-test data. On Hindi speech its absolute score is less trustworthy, but it's still reasonable for *relative* comparisons (e.g. "does the angry-tagged clip sound worse/more strained than the neutral one").
- WER/CER need a transcript to compare against; if your emotion tag makes the model add filler, breathiness, or repeated words in a way that *is* the intended emotional performance, that will show up as a higher WER even though the synthesis is doing exactly what you asked. Read the ASR hypothesis column, not just the number.

```python
import time, json, soundfile as sf
from IPython.display import Audio, display

OUT_DIR = "dhvaani_samples"
os.makedirs(OUT_DIR, exist_ok=True)
sr = dh.sampling_rate

# Rows 1-2 mirror the Indic-MIO neutral baselines exactly -> direct comparison.
# Rows 3-6 are the multilingual medical set from your original DhVaani notebook.
TESTS = [
    ("hi_neutral", "hi", "आज मैंने कुछ ऐसा देखा जिस पर मुझे यकीन ही नहीं हुआ।"),
    ("en_neutral", "en", "I just found out what really happened last night."),
    ("hindi_med",   "hi", "मधुमेह के रोगियों को अपने आहार में मीठे और तले हुए पदार्थों से बचना चाहिए, रोज़ थोड़ा व्यायाम करना चाहिए और समय-समय पर अपने रक्त शर्करा की जाँच करवानी चाहिए।"),
    ("marathi_med", "mr", "मधुमेह असलेल्या रुग्णांनी गोड आणि तळलेले पदार्थ टाळावेत, दररोज थोडा व्यायाम करावा आणि नियमितपणे आपल्या रक्तातील साखरेची तपासणी करून घ्यावी."),
    ("tamil_med",   "ta", "நீரிழிவு நோயாளிகள் இனிப்பு மற்றும் எண்ணெயில் பொரித்த உணவுகளைத் தவிர்க்க வேண்டும், தினமும் சிறிது உடற்பயிற்சி செய்ய வேண்டும், மேலும் அவ்வப்போது தங்கள் இரத்த சர்க்கரை அளவை பரிசோதித்துக் கொள்ள வேண்டும்."),
    ("telugu_med",  "te", "మధుమేహ రోగులు తీపి మరియు నూనెలో వేయించిన ఆహారాలకు దూరంగా ఉండాలి, ప్రతిరోజూ కొంత వ్యాయామం చేయాలి, మరియు అప్పుడప్పుడు తమ రక్తంలో చక్కెర స్థాయిని పరీక్షించుకోవాలి."),
]

manifest, timings, ok, failed = [], {}, 0, 0

for name, lang, text in TESTS:
    try:
        t0 = time.time()
        audio = dh.synthesize(text=text, prompt_wav=REFERENCE_WAV, prompt_text=PROMPT_TEXT)
        synth_seconds = time.time() - t0

        path = os.path.join(OUT_DIR, f"{name}.wav")
        sf.write(path, audio, sr)
        audio_seconds = len(audio) / sr
        rtf = synth_seconds / audio_seconds if audio_seconds > 0 else float("nan")

        print(f"[ok] {name}: {audio_seconds:.2f}s, synth {synth_seconds:.2f}s, RTF {rtf:.2f}")
        display(HTML(f"<b>{name}</b> · {lang}<br>{text}")); display(Audio(path))

        manifest.append({"file": name, "language": lang, "text": text, "emotion": "neutral"})
        timings[name] = {"synth_seconds": synth_seconds,
                         "audio_seconds": audio_seconds, "rtf": rtf}
        ok += 1
    except Exception as e:
        print(f"[FAILED] {name}: {e}")
        failed += 1

for fn, obj in (("manifest.json", manifest), ("timings.json", timings)):
    with open(os.path.join(OUT_DIR, fn), "w", encoding="utf-8") as f:
        json.dump(obj, f, ensure_ascii=False, indent=2)

print(f"\n[done] {ok} ok, {failed} failed -> ./{OUT_DIR}/")
```

```python
!pip install -q jiwer
import os, json, re, torch, jiwer, pandas as pd
from transformers import pipeline

ENGINES = {"dhvaani": "dhvaani_samples", "indic-mio": "indic_mio_samples"}

# second scorer — cross-checks SraVaani so WER isn't hostage to one ASR
whisper = pipeline("automatic-speech-recognition",
                   model="openai/whisper-large-v3-turbo",
                   torch_dtype=torch.float16, device=0, chunk_length_s=30)

def asr_whisper(path, lang):
    return whisper(path, generate_kwargs={"language": lang, "task": "transcribe"})["text"]

def norm(s):
    return re.sub(r"[।,.!?\s]+", " ", s).strip().lower()

rows = []
for engine, d in ENGINES.items():
    manifest = json.load(open(os.path.join(d, "manifest.json"), encoding="utf-8"))
    timings  = json.load(open(os.path.join(d, "timings.json"),  encoding="utf-8"))

    for m in manifest:
        wav = os.path.join(d, m["file"] + ".wav")
        if not os.path.exists(wav):
            print(f"[skip] missing {wav}")
            continue

        ref = norm(m["text"])
        hyp_s = asr(wav)                              # SraVaani
        hyp_w = asr_whisper(wav, m["language"])       # Whisper
        t = timings.get(m["file"], {})

        row = {
            "engine": engine,
            "file": m["file"],
            "language": m["language"],
            "emotion": m.get("emotion", "neutral"),
            "wer_sravaani": jiwer.wer(ref, norm(hyp_s)),
            "cer_sravaani": jiwer.cer(ref, norm(hyp_s)),
            "wer_whisper":  jiwer.wer(ref, norm(hyp_w)),
            "cer_whisper":  jiwer.cer(ref, norm(hyp_w)),
            "rtf": t.get("rtf"),
            "audio_seconds": t.get("audio_seconds"),
            "reference_text": m["text"],
            "asr_sravaani": hyp_s,
            "asr_whisper": hyp_w,
        }
        rows.append(row)
        print(f'{engine:10s} {m["file"]:16s} '
              f'WER sra {row["wer_sravaani"]:.3f} | whi {row["wer_whisper"]:.3f} | '
              f'RTF {row["rtf"] or float("nan"):.2f}')

df = pd.DataFrame(rows)
df.to_csv("tts_eval.csv", index=False, encoding="utf-8-sig")
json.dump(rows, open("tts_eval.json", "w"), ensure_ascii=False, indent=2)
df
```
